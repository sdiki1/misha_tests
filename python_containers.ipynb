а где ответ разошёлся с ожидаемым.

Примеры в условиях записаны так:

```
-> первая строка входных данных
-> вторая строка входных данных

```

Строки, начинающиеся с `->`, программа получает через `input()` — по одному вызову на строку.
Строки после `--`, начинающиеся с `<-`, — ожидаемый вывод.

Проверить решение на своём входе можно так: `from tests import run_with_input` и затем

## 1. `str` — строки

### Как устроены строки

- Строка — **неизменяемая (immutable)** упорядоченная последовательность символов Unicode.
- С Python 3.3 действует «гибкое представление строк» (PEP 393): интерпретатор сам выбирает
  компактный формат хранения (1, 2 или 4 байта на символ) в зависимости от самого «тяжёлого»
  символа в строке. Строка из ASCII-символов занимает меньше памяти, чем строка с эмодзи.
- Раз строки неизменяемы, **любая операция, которая выглядит как «изменение», на деле создаёт
  новый объект** в памяти — исходная строка не трогается.
- Короткие «простые» строки (похожие на идентификаторы) интерпретатор может **интернировать**
  (interning) — кешировать так, что два одинаковых литерала указывают на один и тот же объект.
- Строки **хешируемы** — именно поэтому их можно использовать как ключи словаря или элементы множества.

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | да, порядок символов важен |
| Изменяемость | нет (immutable) |
| Хешируемость | да |
| Индексация/срезы | да |
| Повторяющиеся элементы | допустимы |
| Типы элементов | только символы (строки длины 1) |

````python
s1 = "python"
s2 = "python"
print(s1 is s2, id(s1) == id(s2))   # интернирование простых литералов

s = "python"
print(id(s))
s = s + "!"                          # выглядит как "изменение", но это НОВЫЙ объект
print(id(s))
````

### Операции над строками

````python
s = "python"

# индексация и срезы
print(s[0], s[-1])          # первый и последний символ
print(s[1:4])                # срез
print(s[::-1])                # разворот строки
print(s[::2])                 # каждый второй символ

# конкатенация и повторение
print("py" + "thon")
print("ab" * 3)

# проверка вхождения и сравнение
print("th" in s, "z" in s)
print("abc" < "abd")          # лексикографическое сравнение

# длина, итерация, распаковка
print(len(s))
for ch in s[:3]:
    print(ch, end=" ")
print()
a, b, c, *rest = s
print(a, b, c, rest)
````

### Форматирование строк

````python
name, score = "Михаил", 100

print(f"{name} набрал {score} баллов")          # f-строка (рекомендуемый способ)
print("{} набрал {} баллов".format(name, score))  # .format
print("%s набрал %d баллов" % (name, score))      # старый %-стиль
print(f"{3.14159:.2f}")                            # форматирование числа
print(f"{score:>6}|")                               # выравнивание по правому краю
````

### Основные методы в деле

````python
s = "  Привет, Мир!  "
print(repr(s.strip()))
print(s.strip().lower())
print(s.strip().upper())
print(s.strip().replace("Мир", "Python"))
print("a,b,,c".split(","))
print("-".join(["2026", "09", "24"]))
print("42".isdigit(), "abc".isalpha(), "abc123".isalnum())
print("hello world".title())
print("hello".startswith("he"), "hello".endswith("lo"))
print("hello".find("l"), "hello".index("l"))
print("hello".count("l"))
````

## 2. `list` — списки

### Как устроены списки

- Список — **изменяемый (mutable)** динамический массив: под капотом это непрерывный блок памяти,
  хранящий **указатели на объекты**, а не сами объекты.
- Чтобы `append` работал быстро, CPython выделяет память "с запасом" (over-allocation): при
  нехватке места массив пересоздаётся с некоторым запасом ячеек, поэтому добавление в конец —
  O(1) **амортизированно**, а не всегда мгновенно.
- Вставка/удаление **в начале или середине** требует сдвига всех последующих элементов — O(n).
- Список **не хешируем** — его нельзя использовать как ключ словаря или элемент множества.
- Присваивание `b = a` не копирует список — `a` и `b` указывают на один и тот же объект
  (нужно `a.copy()`, `list(a)` или срез `a[:]` для поверхностной копии).

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | да |
| Изменяемость | да (mutable) |
| Хешируемость | нет |
| Индексация/срезы | да |
| Повторяющиеся элементы | допустимы |
| Типы элементов | любые, можно смешивать |

````python
a = [1, 2, 3]
b = a               # b — это ссылка на тот же объект!
b.append(4)
print(a, b, a is b)

c = a.copy()         # поверхностная копия
c.append(100)
print(a, c)
````

### Операции над списками

````python
lst = [10, 20, 30, 40, 50]

# индексация и срезы
print(lst[0], lst[-1])
print(lst[1:4])
print(lst[::-1])

# изменение через индекс и срез
lst[0] = 99
print(lst)
lst[1:3] = [1000, 2000, 3000]     # срез можно заменить на кусок другой длины
print(lst)
del lst[0]                         # удаление по индексу
print(lst)

# конкатенация и повторение (создают НОВЫЙ список)
x = [1, 2] + [3, 4]
y = [0] * 3
print(x, y)

# += изменяет список на месте (эквивалент extend), в отличие от +
z = [1, 2]
z += [3, 4]
print(z)

# членство, сравнение, агрегаты
nums = [5, 3, 8, 1]
print(3 in nums)
print([1, 2] < [1, 3])             # лексикографическое сравнение
print(len(nums), min(nums), max(nums), sum(nums))

# сортировка: sorted() возвращает новый список, .sort() сортирует на месте
print(sorted(nums))
nums.sort(reverse=True)
print(nums)

# списковые включения (list comprehension)
squares = [x**2 for x in range(6) if x % 2 == 0]
print(squares)

# вложенные списки и поверхностное копирование — частая ловушка
matrix = [[0, 0], [0, 0]]
bad_copy = matrix.copy()
bad_copy[0][0] = 1                  # поверхностная копия НЕ копирует вложенные списки
print(matrix)
````

| Операция | Синтаксис | Пример | Сложность |
|---|---|---|---|

## 3. `tuple` — кортежи

### Как устроены кортежи

- Кортеж — **неизменяемый (immutable)** аналог списка: под капотом это тоже массив указателей
  на объекты, но **фиксированного размера**, выделяемый ровно под нужное число элементов
  (без запаса, в отличие от списка).
- Раз размер и содержимое не меняются, кортеж **хешируем** — но только если хешируемы все его
  элементы (кортеж, содержащий список, хешировать нельзя).
- Из-за неизменяемости и отсутствия over-allocation кортежи немного компактнее и быстрее
  создаются, чем списки той же длины — их удобно использовать для «связок» данных, которые
  не должны меняться (например, координаты точки, запись базы данных).
- Кортеж из одного элемента требует запятой: `(1,)`, а не `(1)` (это просто число в скобках).

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | да |
| Изменяемость | нет (immutable) |
| Хешируемость | да, если хешируемы все элементы |
| Индексация/срезы | да |
| Повторяющиеся элементы | допустимы |
| Типы элементов | любые, можно смешивать |

````python
point = (3, 4)
print(hash(point))                 # кортеж из хешируемых элементов — хешируем
d = {point: "точка A"}             # поэтому кортеж можно использовать как ключ словаря
print(d)

bad = ([1, 2], 3)
try:
    hash(bad)                       # а вот кортеж со списком внутри — нельзя
except TypeError as e:
    print("Ошибка:", e)
````

### Операции над кортежами

````python
t = (10, 20, 30, 40)

# индексация и срезы — как у списка
print(t[0], t[-1], t[1:3])

# распаковка (unpacking) — очень частый паттерн для кортежей
x, y, z, w = t
print(x, y, z, w)
first, *middle, last = t
print(first, middle, last)

# упаковка нескольких значений в кортеж
def min_max(nums):
    return min(nums), max(nums)     # возвращает кортеж (min, max)

lo, hi = min_max([5, 1, 9, 3])
print(lo, hi)

# конкатенация и повторение — создают новый кортеж
print((1, 2) + (3, 4))
print((0,) * 3)

# сравнение и сортировка — как у списков, лексикографически
print((1, 2) < (1, 3))
print(sorted([(2, "b"), (1, "z"), (1, "a")]))   # сортировка по элементам кортежа по очереди

# конвертация между list и tuple
lst = list(t)
lst.append(50)
t2 = tuple(lst)
print(t2)
````

### Шпаргалка: методы `tuple`

У кортежа всего два метода — из-за неизменяемости методов, меняющих содержимое, попросту нет.

| Метод | Что делает | Пример | Результат |
|---|---|---|---|
| `t.count(x)` | сколько раз встречается значение | `(1,1,2).count(1)` | `2` |
| `t.index(x)` | индекс первого вхождения | `(5,6,7).index(6)` | `1` |

### Шпаргалка: операции над `tuple`

| Операция | Синтаксис | Пример | Сложность |
|---|---|---|---|
| Индексация | `t[i]` | `(1,2,3)[0]` → `1` | O(1) |
| Срез | `t[a:b:c]` | `(1,2,3,4)[1:3]` → `(2,3)` | O(k) |
| Конкатенация | `t1 + t2` | `(1,)+(2,)` → `(1,2)` | O(n+m) |
| Повторение | `t * n` | `(1,)*3` → `(1,1,1)` | O(n·k) |
| Членство | `x in t` | — | O(n) |
| Распаковка | `a, b = t` | — | O(n) |
| Длина | `len(t)` | — | O(1) |
| Хеш (если применимо) | `hash(t)` | — | O(n) |

## 4. `dict` — словари

### Как устроены словари

- Словарь — **изменяемая** коллекция пар «ключ → значение», основанная на **хеш-таблице**.
- Ключ хешируется функцией `hash()`, по хешу вычисляется позиция в таблице — поэтому доступ,
  вставка и удаление по ключу работают в среднем за **O(1)**, а не за O(n), как поиск в списке.
- Именно поэтому **ключи должны быть хешируемыми** (и, как следствие, обычно неизменяемыми):
  подходят `str`, `int`, `tuple` (из хешируемых элементов), но не подходят `list`, `dict`, `set`.
- С Python 3.6 (и официально с 3.7) словарь устроен как «компактный dict»: отдельно хранится
  плотный массив записей в порядке добавления и отдельная разреженная хеш-таблица с индексами
  в этот массив. Побочный эффект — **словарь сохраняет порядок добавления ключей**, это гарантия
  языка, а не случайность.
- Значения могут быть любыми объектами, в том числе изменяемыми.

### Операции над словарями

````python
student = {"name": "Михаил", "faculty": "ИИ", "year": 1}

# доступ, безопасный доступ, изменение, удаление
print(student["name"])
print(student.get("age", "не указано"))    # get не бросает ошибку, если ключа нет
student["year"] = 2
student["age"] = 18
del student["age"]
print(student)

# проверка наличия ключа (`in` проверяет именно ключи, не значения)
print("faculty" in student, "Михаил" in student)

# итерация: по ключам, значениям, парам
for key in student:
    print(key, "->", student[key])
for key, value in student.items():
    print(f"{key}: {value}")

# объединение словарей
defaults = {"lang": "ru", "year": 1}
override = {"year": 2, "active": True}
merged_old = {**defaults, **override}      # через распаковку
merged_new = defaults | override           # оператор | (Python 3.9+)
print(merged_old)
print(merged_new)

d3 = dict(defaults)
d3.update(override)                         # update изменяет на месте
print(d3)

# словарные включения (dict comprehension)
squares = {x: x**2 for x in range(5)}
print(squares)

# сравнение словарей — по содержимому, порядок не важен
print({"a": 1, "b": 2} == {"b": 2, "a": 1})
````

## 5. `set` — множества

### Как устроены множества

- Множество — **изменяемая** неупорядоченная коллекция **уникальных** хешируемых элементов,
  тоже построенная на хеш-таблице (как словарь, только без значений — по сути «словарь, у
  которого есть только ключи»).
- Благодаря хеш-таблице операции добавления, удаления и проверки на вхождение в среднем
  выполняются за **O(1)** — именно поэтому `x in some_set` намного быстрее, чем `x in some_list`
  на больших объёмах данных.
- В отличие от `dict`, у `set` **нет** компактного представления с сохранением порядка — порядок
  элементов не гарантирован и может выглядеть «случайным».
- Есть неизменяемый «близнец» — `frozenset`: как `set`, но без методов изменения, а значит
  хешируем и годится как ключ словаря или элемент другого множества.
- Элементы множества обязаны быть хешируемыми (нельзя положить список внутрь множества).

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | нет |
| Изменяемость | да (`set`) / нет (`frozenset`) |
| Хешируемость элементов | обязательна |
| Уникальность элементов | да, дубликаты автоматически схлопываются |
| Проверка вхождения | в среднем O(1) |

````python
nums = {1, 2, 2, 3, 3, 3}      # дубликаты схлопываются сами
print(nums)

try:
    bad = {[1, 2], 3}          # список внутри множества — нельзя, он не хешируем
except TypeError as e:
    print("Ошибка:", e)

fs = frozenset([1, 2, 3])
d = {fs: "неизменяемое множество как ключ"}   # а frozenset — можно
print(d)
````

### Операции над множествами

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# теоретико-множественные операции (создают новое множество)
print(a | b)            # объединение
print(a & b)             # пересечение
print(a - b)              # разность: есть в a, но нет в b
print(a ^ b)               # симметрическая разность: есть ровно в одном из множеств

# те же операции методами (полезно, если аргумент — не множество, а любой iterable)
print(a.union([10, 11]))
print(a.intersection([2, 3, 100]))

# операции "на месте" — изменяют исходное множество
c = {1, 2, 3}
c |= {4, 5}     # эквивалент c.update({4, 5})
print(c)
c &= {1, 4, 5}
print(c)

# отношения между множествами
print({1, 2} <= {1, 2, 3})       # подмножество
print({1, 2, 3} >= {1, 2})        # надмножество
print({1, 2}.isdisjoint({3, 4}))   # нет общих элементов

# членство и включения
print(3 in a)
unique_letters = {ch for ch in "abracadabra"}   # включение множества (set comprehension)
print(unique_letters)

# типичное применение: быстрая дедупликация и пересечение списков
list1 = [1, 2, 2, 3, 4]
print(list(set(list1)))          # уникальные элементы без сохранения порядка
````

````python
from tests import test_task32

def task32():
    first = input()
    second = input()

    def letter_counts(s):
        counts = {}
        for ch in s.lower():
            if ch.isalpha():
                counts[ch] = counts.get(ch, 0) + 1
        return counts

    print("YES" if letter_counts(first) == letter_counts(second) else "NO")

test_task32(task32)
````

**Решение 33.** Множество `seen` хранит значения ключа, которые уже встречались; запись добавляется в результат, только если значение встречено впервые — так сохраняется порядок.

````python
from tests import test_task33

def task33():
    fields = input().split()
    key = input()
    n = int(input())
    records = []
    for _ in range(n):
        values = input().split()
        records.append(dict(zip(fields, values)))
    seen = set()
    result = []
    for record in records:
        if record[key] not in seen:
            seen.add(record[key])
            result.append(record)
    for record in result:
        print(*(record[field] for field in fields))

test_task33(task33)
````

**Решение 34.** Словарь смежности хранит множество соседей для каждой вершины (ребро добавляется в обе стороны). Соседи соседей дают вершины на расстоянии два; объединяем оба множества и убираем стартовую вершину.

````python
from tests import test_task34

def task34():
    m = int(input())
    edges = []
    for _ in range(m):
        u, v = input().split()
        edges.append((u, v))
    start = input()
    adjacency = {}
    for u, v in edges:
        adjacency.setdefault(u, set()).add(v)
        adjacency.setdefault(v, set()).add(u)
    one_step = adjacency.get(start, set())
    two_steps = set()
    for neighbor in one_step:
        two_steps |= adjacency.get(neighbor, set())
    reachable = (one_step | two_steps) - {start}
    print(*sorted(reachable))

test_task34(task34)
````

**Решение 35.** `shift_text` сдвигает код каждой латинской буквы по модулю 26 относительно базы своего регистра; дешифрование — тот же сдвиг со знаком минус. Взлом перебирает 26 сдвигов и выбирает тот, у которого частоты букв расшифровки ближе всего к `ENGLISH_FREQ` (минимальная сумма модулей разностей).

````python
from tests import test_task35

# эталонные частоты букв английского языка, в процентах (приблизительно)
ENGLISH_FREQ = {
    'e': 12.7, 't': 9.1, 'a': 8.2, 'o': 7.5, 'i': 7.0, 'n': 6.7,
    's': 6.3, 'h': 6.1, 'r': 6.0, 'd': 4.3, 'l': 4.0, 'c': 2.8,
    'u': 2.8, 'm': 2.4, 'w': 2.4, 'f': 2.2, 'g': 2.0, 'y': 2.0,
    'p': 1.9, 'b': 1.5, 'v': 1.0, 'k': 0.8, 'j': 0.15, 'x': 0.15,
    'q': 0.10, 'z': 0.07,
}

def task35():
    mode = input()
    if mode == "crack":
        text = input()
    else:
        shift = int(input())
        text = input()

    def shift_text(text, shift):
        result = []
        for ch in text:
            if "a" <= ch <= "z" or "A" <= ch <= "Z":
                base = ord("A") if ch.isupper() else ord("a")
                result.append(chr((ord(ch) - base + shift) % 26 + base))
            else:
                result.append(ch)
        return "".join(result)

    def distance_from_english(text):
        counts = {letter: 0 for letter in ENGLISH_FREQ}
        for ch in text.lower():
            if ch in counts:
                counts[ch] += 1
        total = sum(counts.values())
        return sum(abs(counts[letter] / total * 100 - expected)
                   for letter, expected in ENGLISH_FREQ.items())

    if mode == "encrypt":
        print(shift_text(text, shift))
    elif mode == "decrypt":
        print(shift_text(text, -shift))
    else:
        best_shift = min(range(26), key=lambda s: distance_from_english(shift_text(text, -s)))
        print(best_shift)
        print(shift_text(text, -best_shift))

test_task35(task35)
````
