## NumPy: массивы чисел

### Массив — коробка с числами

Массив — это удобная коробка с числами. В неё складывают числа одного вида, и дальше с ними работают
все сразу. Массив создают из обычного списка командой `np.array`.

````python
scores = np.array([72, 85, 90, 64])
print(scores)
print(type(scores))
````

### Чем массив отличается от списка

Список и массив ведут себя по-разному. Если умножить список на 2, он повторится два раза. Если
умножить массив на 2, удвоится каждое число — именно это обычно и нужно, когда работаешь с оценками
или ценами.

````python
numbers = [1, 2, 3]
print(numbers * 2)            # список повторился

array = np.array([1, 2, 3])
print(array * 2)              # каждое число умножилось
````

### Создание чисел

Числа не всегда набирают вручную. `np.arange(start, stop)` даёт числа от `start` до `stop`, не включая
`stop` — как знакомый `range`. `np.linspace(start, stop, n)` делит отрезок на `n` точек, включая оба
конца. `np.zeros(n)` и `np.ones(n)` создают `n` нулей или единиц — удобная заготовка, которую потом
заполняют.

````python
print(np.arange(0, 10))        # 0, 1, ..., 9
print(np.arange(5, 8))
print(np.linspace(0, 1, 5))    # 5 точек от 0 до 1 включительно
print(np.zeros(4))
print(np.ones(3))
````

### Размер и форма

`size` — сколько всего чисел в массиве. `shape` — форма: для обычного массива это `(число элементов,)`,
а для таблицы чисел — `(строки, столбцы)`.

````python
scores = np.array([72, 85, 90, 64])
print(scores.size, scores.shape)

table = np.array([[90, 80],
                  [70, 60],
                  [85, 95]])
print(table.size, table.shape)     # 6 чисел: 3 строки и 2 столбца
````

### Индексы и срезы

Элементы нумеруются с нуля, как в списке. `a[0]` — первый, `a[-1]` — последний, `a[1:4]` — элементы
с номерами 1, 2, 3 (четвёртый не входит). В двумерном массиве указывают строку и столбец через запятую:
`matrix[0, 1]` — строка 0, столбец 1. А `matrix[0]` — вся первая строка.

````python
scores = np.array([72, 85, 90, 64, 58])
print(scores[0], scores[-1])
print(scores[1:4])

matrix = np.array([[90, 80, 70],
                   [65, 75, 85]])
print(matrix[0, 1])      # строка 0, столбец 1
print(matrix[1])         # вся вторая строка
````

### Действия со всеми числами сразу

Арифметика применяется к каждому числу массива. Прибавить 5 ко всем оценкам — просто `scores + 5`.
Два массива одинаковой длины складываются поэлементно: первое число с первым, второе со вторым.

````python
scores = np.array([72, 85, 90, 64])
print(scores + 5)
print(scores - 10)
print(scores * 2)
print(scores / 100)
print(scores ** 2)

bonus = np.array([3, 0, 5, 10])
print(scores + bonus)    # поэлементно: 72+3, 85+0, 90+5, 64+10
````

### Частые функции

Для всего массива сразу: `np.sum` — сумма, `np.mean` — среднее, `np.min` и `np.max` — наименьшее
и наибольшее число, `np.round(a, 1)` — округлить каждое число до одного знака после точки.

````python
scores = np.array([72, 85, 90, 64])
print(np.sum(scores))
print(np.mean(scores))
print(np.min(scores), np.max(scores))
print(np.round(scores / 3, 1))
````

### Условия и маски

Сравнение с числом тоже применяется ко всем элементам. `scores >= 70` даёт массив из `True` и `False` —
его называют **маской**. Если поставить маску в квадратные скобки, останутся только те числа, где стоит
`True`. А `np.sum(маска)` считает, сколько там `True`: при сложении `True` считается за 1.

````python
scores = np.array([72, 85, 90, 64, 58])
mask = scores >= 70
print(mask)
print(scores[mask])
print(scores[scores >= 70])      # то же самое, но короче
print(np.sum(scores >= 70))      # сколько оценок не ниже 70
````

### Двумерный массив и `axis`

Пусть в строках — ученики, а в столбцах — предметы. `np.mean(table)` даст одно число — среднее по всем
клеткам. Чтобы получить среднее **по каждому предмету**, нужно пройти по строкам сверху вниз и получить
по числу на каждый столбец — это `axis=0`. Чтобы получить среднее **по каждому ученику**, идём по строке
слева направо — это `axis=1`.

Как запомнить: `axis=0` — ответ длиной в число столбцов, `axis=1` — в число строк.

````python
#                 математика  python
table = np.array([[90, 80],     # Аня
                  [70, 60],     # Боря
                  [85, 95]])    # Вера
print(np.mean(table))             # одно число — среднее по всем клеткам
print(np.mean(table, axis=0))     # по предметам: столбец за столбцом
print(np.mean(table, axis=1))     # по ученикам: строка за строкой
print(np.max(table, axis=0))      # лучший балл по каждому предмету
````

### Выбор строк: `iloc` и `loc`

`iloc` выбирает строки **по номеру**: `df.iloc[0]` — первая строка, `df.iloc[0:3]` — первые три.

`loc` выбирает **по метке строки и имени столбца**: `df.loc[2, "name"]` — имя из строки с меткой 2.
Чаще всего `loc` используют с условием: `df.loc[df["city"] == "Казань", ["name", "city"]]` — строки,
где город Казань, и только два столбца из них.

````python
print(df.iloc[0])                 # первая строка целиком
df.iloc[0:3]
````

````python
print(df.loc[2, "name"])
df.loc[df["city"] == "Казань", ["name", "city", "math_score"]].head()
````

### Фильтрация

Условие на столбец даёт маску — столбец из `True` и `False`, как в NumPy. Маска в квадратных скобках
оставляет только нужные строки. Два условия соединяют знаком `&` («и»), и **каждое условие берут
в скобки** — без скобок Python прочитает запись неправильно.

````python
moscow = df[df["city"] == "Москва"]
print(moscow.shape)
moscow.head()
````

````python
strong = df[(df["grade"] == 11) & (df["math_score"] >= 80)]
print("одиннадцатиклассников с математикой от 80:", len(strong))
strong.head()
````

### Новый столбец

Новый столбец создают присваиванием: имя в квадратных скобках слева, значения справа. Средний балл
по двум предметам — `df[["math_score", "python_score"]].mean(axis=1)`: среднее по каждой строке.
Если один из баллов пропущен, pandas не считает пропуск нулём — средним станет оставшийся балл.

````python
df["average_score"] = df[["math_score", "python_score"]].mean(axis=1)
df[["name", "math_score", "python_score", "average_score"]].head()
````

### Сортировка

`sort_values("столбец")` сортирует по возрастанию, а с `ascending=False` — по убыванию. Исходная
таблица при этом не меняется: результат — новая таблица, которую можно сохранить в переменную.

````python
df.sort_values("python_score", ascending=False).head()
````

### Подсчёт категорий

`value_counts()` считает, сколько раз встречается каждое значение в столбце. Так узнают, сколько
учеников из каждого города или в каждом классе.

````python
print(df["city"].value_counts())
print(df["grade"].value_counts())
````

### Пропуски

В таблицах бывают пустые клетки — **пропуски**. pandas показывает их как `NaN` («not a number», не число).
Пропуск — это *неизвестное* значение, а не ноль: при подсчёте среднего pandas его просто не учитывает,
а сравнение с пропуском всегда даёт `False`.

`isna()` показывает, где стоят пропуски; `isna().sum()` — сколько их в каждом столбце. `fillna(значение)`
заменяет пропуски на указанное значение — чаще всего на медиану или среднее столбца. Чтобы не портить
исходную таблицу, сначала делают копию: `df.copy()`.

````python
print(df.isna().sum())
print(df["math_score"].isna().head())
````

````python
filled = df.copy()
median_math = filled["math_score"].median()
print("медиана по математике:", median_math)
filled["math_score"] = filled["math_score"].fillna(median_math)
print("пропусков в копии:", filled["math_score"].isna().sum(), "| в исходной таблице:", df["math_score"].isna().sum())
````

### Группировка

`groupby("city")` разбивает таблицу на группы по городу, а следующая команда считает что-то внутри каждой
группы. `mean()` — средний балл по каждому городу, `count()` — сколько заполненных значений в группе.

````python
print(df.groupby("city")["math_score"].mean().round(1))
print(df.groupby("grade")["python_score"].mean().round(1))
print(df.groupby("grade")["math_score"].count())
````

### Вопрос к данным

Теперь можно отвечать на вопросы. Например: в каком классе средний балл по Python самый высокий?
Группируем по классу, считаем среднее, сортируем по убыванию и смотрим первую строку.

````python
by_grade = df.groupby("grade")["python_score"].mean().round(1)
print(by_grade.sort_values(ascending=False).head(1))
````

**Решение 6. square_numbers.** `**` возводит в квадрат каждое число.

**Решение 16. add_average_score.** Сначала `copy()`, затем новый столбец через `mean(axis=1)` — пропуск не считается нулём.

````python
from tests_numpy_pandas import test_task16

def add_average_score(df):
    result = df.copy()
    result["average_score"] = result[["math_score", "python_score"]].mean(axis=1)
    return result

test_task16(add_average_score)
````

**Решение 18. students_per_city.** `value_counts` считает, сколько раз встретился каждый город.

**Решение 19. fill_math_scores.** Медиана считается без пропусков, `fillna` заменяет их на неё — в копии таблицы.

````python
from tests_numpy_pandas import test_task19

def fill_math_scores(df):
    result = df.copy()
    median = result["math_score"].median()
    result["math_score"] = result["math_score"].fillna(median)
    return result

test_task19(fill_math_scores)
````

**Решение 20. mean_score_by_grade.** `groupby("grade")` делит таблицу на классы, `mean()` считает среднее в каждом.

````python
from tests_numpy_pandas import test_task20

def mean_score_by_grade(df):
    return df.groupby("grade")["math_score"].mean()

test_task20(mean_score_by_grade)
````
